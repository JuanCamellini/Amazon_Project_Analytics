This notebook represented the initial approach of the project.
It extracted ASINs from the Kaggle dataset to query them individually via the API.

**Why it was discontinued:** The CSV products belonged to Amazon India (prices in ₹),
making them unrepresentative of the target market. This approach was replaced by a
category-based search on amazon.com (US), which provides a more controlled and
representative dataset for analysis.

```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt
import os

BASE_DIR = os.path.abspath("..")

# Loading amazon CSV dataset
sales_df = pd.read_csv(os.path.join(BASE_DIR, "data", "raw", "amazon_sales_dataset.csv"))
```

```python
# Validate ASINs: start with 'B' and has 10 caracters
mascara_validos = (sales_df["product_id"].str.startswith("B")) & (sales_df["product_id"].str.len() == 10)
print(f"Total registros: {len(sales_df)}")
print(f"IDs válidos (ASIN): {mascara_validos.sum()}")
print(f"IDs inválidos: {(~mascara_validos).sum()}")
```

```python
# Get only the ASINs, eliminate duplicates
asins_to_query = sales_df[mascara_validos]["product_id"].drop_duplicates().reset_index(drop=True)

# Export with a column named 'asin'
asins_to_query.to_frame(name="asin").to_csv("../data/raw/asins_to_query.csv", index=False)
print(f"ASINs exportados: {len(asins_to_query)}")
```
